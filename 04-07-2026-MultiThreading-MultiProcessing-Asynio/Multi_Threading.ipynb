### Regular Program

```python
import time
def square_num(num):
    print("Squaring the number.")
    for i in num:
        squared = i ** 2
        time.sleep(0.2)
        print(squared)

def cube_num(num):
    print("Cube the number.")
    for i in num:
        cubed = i ** 3
        time.sleep(0.2)
        print(cubed)

num = [2, 4, 6, 8, 10]

start_time = time.time()

square_num(num)
cube_num(num)

end_time = time.time()

print(f"Total execution time: {end_time - start_time:.4f} seconds")
```

### Program with Threading

```python
import threading

start_time = time.time()

thread1 = threading.Thread(target=square_num, args=(num,))
thread2 = threading.Thread(target=cube_num, args=(num,))

thread1.start()
thread2.start()

thread1.join()
thread2.join()

end_time = time.time()

print(f"Total execution time: {end_time - start_time:.4f} seconds")
```
