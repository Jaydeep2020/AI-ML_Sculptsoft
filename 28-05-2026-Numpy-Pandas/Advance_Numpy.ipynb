```python
import numpy as np
```

## Numpy Stacking
- Stacking means combining multiple arrays together into a single array.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

vertical_join = np.vstack((a,b))  # Joins arrays row-wise (top to bottom).

horizontal_join = np.hstack((a,b))  # Joins arrays column-wise (side by side).

stacked = np.stack((a,b), axis=1) # Adds a new axis while combining arrays.

a = np.array([[1,2]])
b = np.array([[3,4]])

concat = np.concatenate((a,b), axis=1)  # General method to combine arrays.


print(vertical_join)
print()
print(horizontal_join)
print()
print(stacked)
print()
print(concat)
```

## Numpy Splitting
- Splitting means breaking one array into smaller arrays.
- np.split() → Splits an array into multiple sub-arrays.
- np.hsplit() → Splits an array horizontally (column-wise).
- np.vsplit() → Splits an array vertically (row-wise).

```python
a = np.array([1,2,3,4,5,6])
```

```python
result = np.split(a, 3)
print(result)
```

```python
# Horizontal Split

a = np.array([[1,2,3,4],
              [5,6,7,8]])

print(np.hsplit(a, 2))
```

```python
# Vertical Split

a = np.array([[1,2],
              [3,4],
              [5,6],
              [7,8]])

print(np.vsplit(a,2))
```

## Linear Algebra Operations

```python
a = np.array([[1,2],
              [3,4]])

b = np.array([[5,6],
              [7,8]])

print("Matrix Multiplication: \n", a @ b)  # ✅ Best

print(np.dot(a, b))  # ⚠️ Older/general

print(np.matmul(a, b)) # good
```

```python
# Transpose of Matrix

a = np.array([[1,2,3],
              [4,5,6]])

print(a.T)
```

```python
# Determinant of Matrix

a = np.array([[1,2],
              [3,4]])

det = np.linalg.det(a)

print(det)
```

```python
# Inverse of Matrix

print(np.linalg.inv(a))
```

#### Eigen Values & Eigen Vectors

-> Very important in:

- PCA

- Computer Vision

```python
a = np.array([[1,2],
              [3,4]])

values, vectors = np.linalg.eig(a)

print(values)
print()
print(vectors)
```
