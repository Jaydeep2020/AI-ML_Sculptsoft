## Arithmetic Operations

- Subtraction (`-`)

- Modulus (`%`)

## Statistical Operations
- Mean → `np.mean()`
- Median → `np.median()`
- Mode → `stats.mode()`
- Standard Deviation → `np.std()`
- Variance → `np.var()`
- Sum → `np.sum()`
- Minimum → `np.min()`
- Maximum → `np.max()`

---

## Mathematical Functions
- Square Root → `np.sqrt()`
- Absolute Value → `np.abs()`
- Exponential → `np.exp()`
- Logarithm → `np.log()`

---

## Trigonometric Functions
- Sine → `np.sin()`

## Rounding Operations
- Floor → `np.floor()`
- Ceil → `np.ceil()`
- Round → `np.round()`

```python
import numpy as np
```

```python
arr  = np.array([1,2,3,4,6,7,8])
```

```python
# Addition
print(arr + 2)

# Power
print(arr ** 2)

# Modulus
print(arr % 2)
```

```python
# Statistical Operations

arr  = np.array([1,2,3,4,6,4,8,2])

# Mean
mean = np.mean(arr)
print(mean)

# Median
median = np.median(arr)
print(median)
```

```python
# Sum

sum_arr = np.sum(arr)
print(sum_arr)
```

```python
# Log
print("Log: ", np.log(arr) ,"\n")

# Exponential
print("Exponent: ", np.exp(arr))
```

## Matrix Multiplication

```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

print(a @ b)
print()
print(np.matmul(a, b))
```

## Rounding Operations

```python
arr = np.array([1.2, 2.7, 3.5])

# Floor
print("Floor rounding: ", np.floor(arr))

# Ceil
print("Ceil rounding: ", np.ceil(arr))

# Round
print("Round: ", np.round(arr))
```
