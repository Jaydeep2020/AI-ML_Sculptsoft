## Cleaning Data

```python
df[df["Name"].isna()]
```

```python
df["Name"] = df["Name"].fillna("Unknown")  # Null name filled with 'Unknown'

df[df["Name"].isna()]
```

```python
# Department column

df[df["Department"].isna()]
```

```python
df = df.dropna(subset=["Department"])  # Drop rows where Department is null
```

```python
df[df["Department"].isna()]
```

```python
df = df.reset_index(drop=True) # Reset index after dropping rows
```

```python
# cleaning age column
median_age = df["Age"].median()

print(median_age)
print()

df["Age"] = df["Age"].fillna(median_age)
```

```python
# cleaning City column

df["City"].value_counts()
```

```python
most_frequent_city = df["City"].mode()[0]
most_frequent_city
```

```python
df["City"] = df["City"].fillna(most_frequent_city)
```

```python
df.isna().sum()
```

```python
# Final Cleaned Data
print(df.to_string())
```
