## Panda Series
- A Pandas Series is like a column in a table.

- It is a one-dimensional array holding data of any type.

````python
a = [1,2, "apple"]

series = pd.Series(a, index = ["x", "y", "z"])

print(a)
````

````python
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
````

# DataFrames in Pandas

- A Pandas DataFrame is a two-dimensional, tabular data structure in Python that organizes data into rows and columns.

````python
data = {
    "name" : ["Jay", "John", "Mike", "Priya"],
    "age" : [23, 34, None, 19],
    "Married": [False, True, False, False],
    "weight": [65.7, 75.2, 88.9, None]
}

df = pd.DataFrame(data)

print(df.columns)
print()
print(df)
````

# loc vs iloc in Pandas

- Both are used to access rows and columns in a DataFrame.

| Method | Works With |
|---|---|
| `loc[]` | Labels / Names |
| `iloc[]` | Integer Positions |

### Examples

```python
df.loc[0]      # Access using label
df.iloc[0]     # Access using position

````python
df.loc[:, "name"]
````

````python
# Custom rows and columns fetching

df.loc[[0, 2],["name","Married"]]
````

````python
df.iloc[:2, :3]
````

````python
df.iloc[0:1, 0:1]
````

````python
print(df.to_string())
````
