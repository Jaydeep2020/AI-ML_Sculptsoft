## Correlation betweens columns using Pandas

```python
df[["Temperature(°C)", "Dew point temperature(°C)"]].corr()
```

```python
import matplotlib.pyplot as plt

df.plot(kind="scatter", x="Dew point temperature(°C)", y="Temperature(°C)")

plt.show()
```
