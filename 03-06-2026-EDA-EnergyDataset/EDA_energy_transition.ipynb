## Handling Missing Values

#### Replace inf values with NaN

```python
np.isinf(df.select_dtypes(include='number')).sum() # find all 'inf' values
```

```python
# Replace inf with NaN in growth rate columns

growth_cols = ['solar_yoy_growth_pct', 'wind_yoy_growth_pct', 'renewables_yoy_growth_pct']

for col in growth_cols:
    df[col] = df[col].replace([np.inf, -np.inf], np.nan)
```

```python
np.isinf(df.select_dtypes(include='number')).sum()
```

#### Impute missing GDP values with country-specific mean

```python
df['gdp_usd'] = df.groupby('country')['gdp_usd'].transform(
    lambda x: x.fillna(x.mean())
)
```

```python
# --------------------------------------------------
# 1. Wind generation and share
# Countries with no wind generation -> fill with 0
# --------------------------------------------------
no_wind = (
    df.groupby('country')['wind_electricity_twh']
    .transform('sum')
    .fillna(0)
    .eq(0)
)

df.loc[no_wind & df['wind_electricity_twh'].isna(),
           'wind_electricity_twh'] = 0

df.loc[no_wind & df['wind_share_pct'].isna(),
           'wind_share_pct'] = 0
```

```python
# --------------------------------------------------
# 2. Hydro & Nuclear generation
# Interpolate within each country over time
# --------------------------------------------------
for col in ['hydro_electricity_twh', 'nuclear_electricity_twh']:
    df[col] = (
        df.sort_values('year')
        .groupby('country')[col]
        .transform(lambda x: x.interpolate(limit_direction='both'))
    )
```

```python
# --------------------------------------------------
# 3. Growth columns
# Calculate from previous year if possible
# --------------------------------------------------
for col in [
    'solar_yoy_growth_pct',
    'wind_yoy_growth_pct',
    'renewables_yoy_growth_pct'
]:
    df[col] = (
        df.groupby('country')[col]
        .transform(lambda x: x.interpolate(limit_direction='both'))
    )
```

```python
# Remaining missing values -> median
growth_cols = [
    'solar_yoy_growth_pct',
    'wind_yoy_growth_pct',
    'renewables_yoy_growth_pct'
]

for col in growth_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# --------------------------------------------------
# 4. Policy milestone
# Missing means no recorded policy
# --------------------------------------------------
df['policy_milestone'] = (
    df['policy_milestone']
    .fillna('No recorded policy')
)
```

```python
df['hydro_electricity_twh'] = (
    df.groupby('country')['hydro_electricity_twh']
      .transform(lambda x: x.interpolate(limit_direction='both'))
)
```
