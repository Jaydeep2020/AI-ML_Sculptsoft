## Handling Missing Values

#### Replace inf values with NaN

```python
np.isinf(df.select_dtypes(include='number')).sum() # find all 'inf' values
```

```python
# Replace inf with NaN in growth rate columns

growth_cols = ['solar_yoy_growth_pct', 'wind_yoy_growth_pct', 'renewables_yoy_growth_pct']

for col in growth_cols:
    df[col] = df[col].replace([np.inf, -np.inf], np.nan)
```

```python
np.isinf(df.select_dtypes(include='number')).sum()
```

#### Impute missing GDP values with country-specific mean

```python
df['gdp_usd'] = df.groupby('country')['gdp_usd'].transform(
    lambda x: x.fillna(x.mean())
)
```

```python
# --------------------------------------------------
# 1. Wind generation and share
# Countries with no wind generation -> fill with 0
# --------------------------------------------------
no_wind = (
    df.groupby('country')['wind_electricity_twh']
    .transform('sum')
    .fillna(0)
    .eq(0)
)

df.loc[no_wind & df['wind_electricity_twh'].isna(),
           'wind_electricity_twh'] = 0

df.loc[no_wind & df['wind_share_pct'].isna(),
           'wind_share_pct'] = 0
```

```python
# --------------------------------------------------
# 2. Hydro & Nuclear generation
# Interpolate within each country over time
# --------------------------------------------------
for col in ['hydro_electricity_twh', 'nuclear_electricity_twh']:
    df[col] = (
        df.sort_values('year')
        .groupby('country')[col]
        .transform(lambda x: x.interpolate(limit_direction='both'))
    )
```

```python
# --------------------------------------------------
# 3. Growth columns
# Calculate from previous year if possible
# --------------------------------------------------
for col in [
    'solar_yoy_growth_pct',
    'wind_yoy_growth_pct',
    'renewables_yoy_growth_pct'
]:
    df[col] = (
        df.groupby('country')[col]
        .transform(lambda x: x.interpolate(limit_direction='both'))
    )
```

```python
# Remaining missing values -> median
growth_cols = [
    'solar_yoy_growth_pct',
    'wind_yoy_growth_pct',
    'renewables_yoy_growth_pct'
]

for col in growth_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# --------------------------------------------------
# 4. Policy milestone
# Missing means no recorded policy
# --------------------------------------------------
df['policy_milestone'] = (
    df['policy_milestone']
    .fillna('No recorded policy')
)
```

```python
df['hydro_electricity_twh'] = (
    df.groupby('country')['hydro_electricity_twh']
      .transform(lambda x: x.interpolate(limit_direction='both'))
)
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# ── Global aggregated generation mix over time ───────────────────────────────
global_yr = df.groupby('year').agg(
    total_gen   = ('total_electricity_generation_twh', 'sum'),
    solar_gen   = ('solar_electricity_twh',            'sum'),
    wind_gen    = ('wind_electricity_twh',             'sum'),
    hydro_gen   = ('hydro_electricity_twh',            'sum'),
    nuclear_gen = ('nuclear_electricity_twh',          'sum'),
    fossil_gen  = ('fossil_electricity_twh',           'sum'),
    co2_saved   = ('co2_saved_solar_wind_mt',          'sum'),
    avg_ci      = ('carbon_intensity_gco2_kwh',        'mean'),
).reset_index()

# Compute shares
for src in ['solar', 'wind', 'hydro', 'nuclear', 'fossil']:
    global_yr[f'{src}_share'] = global_yr[f'{src}_gen'] / global_yr['total_gen'] * 100

# ── Define Palette ───────────────────────────────────────────────────────────
TECH_PALETTE = {
    'Solar': '#ffb703',
    'Wind': '#8ecae6',
    'Hydro': '#219ebc',
    'Nuclear': '#9b5de5',
    'Fossil': '#4a5759'
}

# ── Plotting ─────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('Global Electricity Mix: 25-Year Transformation (2000–2025)', fontsize=15, fontweight='bold')

# — Stacked area: generation mix
area_data = global_yr[['solar_gen','wind_gen','hydro_gen','nuclear_gen','fossil_gen']].values.T
labels = ['Solar', 'Wind', 'Hydro', 'Nuclear', 'Fossil']
colors = [TECH_PALETTE[l] for l in labels]

axes[0,0].stackplot(global_yr['year'], area_data, labels=labels, colors=colors, alpha=0.88)
axes[0,0].set_title('Electricity Generation Mix (TWh)')
axes[0,0].set_ylabel('TWh')
axes[0,0].legend(loc='upper left', fontsize=8, ncol=2)
axes[0,0].xaxis.set_major_locator(mticker.MultipleLocator(5))

# — Share trends: solar, wind, fossil
ax2 = axes[0,1]
ax2.plot(global_yr['year'], global_yr['fossil_share'], color=TECH_PALETTE['Fossil'],
         linewidth=2.2, label='Fossil Share %', marker='o', markersize=3)
ax2.plot(global_yr['year'], global_yr['solar_share'], color=TECH_PALETTE['Solar'],
         linewidth=2.2, label='Solar Share %', marker='s', markersize=3)
ax2.plot(global_yr['year'], global_yr['wind_share'], color=TECH_PALETTE['Wind'],
         linewidth=2.2, label='Wind Share %', marker='^', markersize=3)
ax2.set_title('Fossil vs Solar vs Wind — Share of Generation (%)')
ax2.set_ylabel('%')
ax2.legend(fontsize=9)
ax2.xaxis.set_major_locator(mticker.MultipleLocator(5))

# — CO2 savings (Added 'r' before strings to fix syntax warnings)
ax3 = axes[1,0]
ax3.fill_between(global_yr['year'], global_yr['co2_saved'],
                 color='#2d6a4f', alpha=0.35, label=r'$\mathregular{CO_2}$ Saved (Mt)')
ax3.plot(global_yr['year'], global_yr['co2_saved'], color='#1b4332', linewidth=2)
ax3.set_title(r'Cumulative $\mathregular{CO_2}$ Avoided by Solar & Wind (Mt/year)')
ax3.set_ylabel(r'Mt $\mathregular{CO_2}$')
ax3.xaxis.set_major_locator(mticker.MultipleLocator(5))

# — Average carbon intensity (Added 'r' before string to fix syntax warning)
ax4 = axes[1,1]
ax4.plot(global_yr['year'], global_yr['avg_ci'], color='#e76f51', linewidth=2.2, marker='D', markersize=3)
ax4.fill_between(global_yr['year'], global_yr['avg_ci'], alpha=0.15, color='#e76f51')
ax4.set_title(r'Average Carbon Intensity ($\mathregular{gCO_2}$/kWh)')
ax4.set_ylabel(r'g$\mathregular{CO_2}$/kWh')
ax4.xaxis.set_major_locator(mticker.MultipleLocator(5))

for ax in axes.flat:
    ax.set_xlabel('Year')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

# ── Compare renewables growth rate: policy year vs non-policy year ───────────
policy_effect = df.copy()
policy_effect['milestone_year'] = policy_effect['has_policy_milestone']

# Average growth rate: milestone year vs all other years
growth_comp = (policy_effect.groupby('milestone_year')['renewables_yoy_growth_pct']
               .agg(['mean','median','count'])
               .rename(index={0:'Non-Milestone Year', 1:'Milestone Year'}).round(2))
print("Growth Rate Comparison: Milestone vs Non-Milestone Years")
print(growth_comp)
print()

# ── Color Palette Definitions ────────────────────────────────────────────────
COLOR_MILESTONE = '#2d6a4f'      # Forest Green (Represents policy progression)
COLOR_NON_MILESTONE = '#e76f51'  # Coral/Terracotta (Represents non-policy / control)

# Distribution plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Policy Milestones & Renewable Energy Acceleration', fontsize=14, fontweight='bold')

for label, grp in policy_effect.groupby('milestone_year'):
    name = 'Milestone Years' if label else 'Non-Milestone Years'
    color = COLOR_MILESTONE if label else COLOR_NON_MILESTONE
    data = grp['renewables_yoy_growth_pct'].dropna()
    data = data[data.between(-30, 100)]
    sns.kdeplot(data, ax=axes[0], label=name, color=color, fill=True, alpha=0.3, linewidth=2)

axes[0].set_title('Renewables YoY Growth: Milestone vs Non-Milestone')
axes[0].set_xlabel('YoY Growth (%)')
axes[0].legend()

# Countries with milestones — show their trajectory
milestone_countries = df[df['has_policy_milestone']==1]['country'].unique()
milestone_traj = df[df['country'].isin(milestone_countries)].groupby('year')['renewables_share_pct'].mean()
non_milestone_traj = df[~df['country'].isin(milestone_countries)].groupby('year')['renewables_share_pct'].mean()

# Unified labels and colors to match the KDE distribution plot
axes[1].plot(milestone_traj.index, milestone_traj.values, color=COLOR_MILESTONE, linewidth=2.2,
             label='Countries with Policy Milestones', marker='o', markersize=3)
axes[1].plot(non_milestone_traj.index, non_milestone_traj.values, color=COLOR_NON_MILESTONE, linewidth=2.2,
             label='Countries without Milestones', marker='s', markersize=3)
axes[1].set_title('Average Renewables Share: Policy Leaders vs Rest')
axes[1].set_ylabel('Avg Renewables Share (%)')
axes[1].set_xlabel('Year')
axes[1].legend()
axes[1].xaxis.set_major_locator(mticker.MultipleLocator(5))

plt.tight_layout()
# plt.savefig('fig14_policy.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
```
