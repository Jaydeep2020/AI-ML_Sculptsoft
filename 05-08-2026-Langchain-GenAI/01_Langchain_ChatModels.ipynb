### > OpenAI LLM Call using Langchain

```python
from dotenv import load_dotenv
import os

from langchain_openai import ChatOpenAI

# load .env
load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

model = ChatOpenAI(
    model='gpt-4.1-mini',
    temperature=0.7
)

response = model.invoke("Explain transformer in 3 simple points.")

print(response)
```

```python
# Display Markdown

from IPython.display import Markdown, display

display(Markdown(response.content))
```
