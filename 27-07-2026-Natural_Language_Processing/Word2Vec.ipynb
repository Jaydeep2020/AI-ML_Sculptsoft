```python
import gensim
import os
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

story = []

for filename in os.listdir('text_data'):

    print(filename)

    with open(os.path.join('text_data', filename), encoding='cp1252') as f:
        corpus = f.read()
        raw_sent = sent_tokenize(corpus)
        for sent in raw_sent:
            story.append(simple_preprocess(sent))
```

```python
corpus[:100]
```

```python
story[1]
```

```python
print("Total Sentences : ", len(story))
```

```python
model = gensim.models.Word2Vec(
    vector_size=100,
    window=10,
    min_count=2,
    workers=4
)
```

- **model.build_vocab(story)** builds the vocabulary from your training data. It scans all the sentences, counts how many times each word appears, and creates the list of words that the Word2Vec model will learn.

```python
model.build_vocab(story)
```

```python
model.epochs
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```
