```python
# Handling Missing Values

missing_value_percentage = df['gender'].isna().mean() * 100

print(missing_value_percentage)

df = df.dropna(subset=['gender'])
```

```python
df['gender'].isna().sum()
```

```python
plt.figure(figsize=(5, 4))
plt.title("User comparison Gender wise")
sns.countplot(df, x='gender')
```

```python
sns.histplot(df, x='age', bins=20, kde=True)

plt.title("User's Age Distribution")
```

```python
df["activity_score"] = (
    df["likes"] +
    df["friendships_initiated"] +
    df["likes_received"]
)

age_activity = df.groupby("age")["activity_score"].mean()

age_activity.sort_values(ascending=False).head(10)

age_activity.plot(kind="line")
plt.xlabel("Age")
plt.ylabel("Average Activity Score")
plt.title("Activity by Age")
plt.show()
```
