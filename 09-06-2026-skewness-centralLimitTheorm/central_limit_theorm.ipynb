# Central Limit Theorem (CLT)

## Definition

The **Central Limit Theorem (CLT)** states that:

> If we repeatedly take random samples from any population and calculate their means, the distribution of those sample means will approach a normal distribution as the sample size increases.

This is true **even if the original population is not normally distributed**.

---

## Easy Formula

$$
\bar{X} \sim N\left(\mu,\frac{\sigma}{\sqrt{n}}\right)
$$

Where:

- **μ** = Population Mean
- **σ** = Population Standard Deviation
- **n** = Sample Size
- **σ / √n** = Standard Error
- **X̄** = Sample Mean

---

## Conditions for CLT

### 1. Random Sampling
Samples should be selected randomly.

### 2. Independent Observations
Each observation should not depend on another.

### 3. Large Sample Size
Generally:

- **n ≥ 30** → CLT works well for most cases.
- Smaller samples may work if the population is already normal.

---

## CLT in Machine Learning

### Used In

- Confidence Intervals
- A/B Testing
- Statistical Inference
- Feature Analysis
- Sampling Techniques
- Model Evaluation

---

Remember:

> **"Means become Normal."**

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Population (highly skewed)
population = np.random.exponential(scale=2, size=100000)
```

```python
sns.kdeplot(population, fill=True)
```

```python
# Parameters
sample_size = 30
num_samples = 1000
```

```python
# Collect sample means

sample_means= []

for _ in range(num_samples):
    sample = np.random.choice(population, size=sample_size)
    sample_means.append(sample.mean())
```

```python
# Plot
plt.figure(figsize=(8,5))
sns.histplot(sample_means, kde=True)
plt.title(f"Distribution of Sample Means (n={sample_size})")
plt.xlabel("Sample Mean")
plt.show()
```
