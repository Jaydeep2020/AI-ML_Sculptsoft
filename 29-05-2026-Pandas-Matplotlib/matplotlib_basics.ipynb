# Matplotlib Basics

- Matplotlib is a low level graph plotting library in python that serves as a visualization utility.

```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.lines import lineStyles
```

```python
# df = pd.read_csv("SeoulBikeData.csv", encoding="cp1252")
df = pd.read_csv("sharma-kohli.csv")
```

```python
df.head()
```

```python
# Lineplot, with title, labels, legend, and grid. Line-color, linestyle

plt.figure(figsize=(8, 4))
plt.plot(df["index"], df["R Sharma"], color="cyan", marker=">", linestyle="dashed", label="R Sharma")
plt.plot(df["index"], df["V Kohli"], color="hotpink", marker="D", linestyle="dashdot", label="V Kohli")
plt.title("Sharma-Kohli")
plt.xlabel("Year")
plt.ylabel("Match Score")
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

```python
df2 = pd.read_csv("SeoulBikeData.csv", encoding="cp1252")
df2.head()
```

```python
plt.title("Rented Bike Count Distribution")
plt.ylabel("Bike Count")

plt.boxplot(df2["Rented Bike Count"], patch_artist=True, boxprops=dict(facecolor="lightblue"), whiskerprops=dict(color="hotpink"),medianprops=dict(color="yellow", linewidth=2),showmeans=True,notch=True, vert=False)

plt.show()
```

```python
# Pie Plot using Matplotlib

holiday_counts = df2["Holiday"].value_counts()

# plt.pie(data, labels=labels)

plt.pie(holiday_counts, labels=holiday_counts.index, autopct="%1.1f%%", textprops={"color": "red", "fontsize": 12})

plt.title("Holiday or Working Day in year Distribution")

plt.show()
```

```python
# Histogram using Matplotlib

plt.figure(figsize=(8, 4))

plt.hist(df2["Temperature(°C)"], bins=10, rwidth=0.6, color="lightblue")

mean = df2["Temperature(°C)"].mean()
plt.axvline(mean, linestyle="dashed", color="hotpink", label=f"Mean: {mean:.2f}")

median = df2["Temperature(°C)"].median()
plt.axvline(median, linestyle="dashed", color="yellow", label=f"Mean: {mean:.2f}")

plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")

plt.show()
```
