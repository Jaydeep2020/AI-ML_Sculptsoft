```python
import pandas as pd
```

### Concatenation of DataFrames in Pandas

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"]
})

df2 = pd.DataFrame({
    "ID": [4, 5, 6],
    "Name": ["David", "Eve", "Frank"]
})

concat_df = pd.concat([df1, df2], ignore_index=True)
print(concat_df)
```

```python
df1.columns
```

## Joins in Pandas

```python
merge_df = pd.merge(df1, df2, on="ID", how="outer")
print(merge_df)
```

```python
df2.columns
```

```python
print(df1.join(df2, lsuffix="_left", rsuffix="_right"))
```

### compare() → Find differences

- Used to compare two DataFrames and show changed values.

- Explanation:

    - Row 1 is different
    - self = value from df1
    - other = value from df2

```python
df1 = pd.DataFrame({
    "marks": [85, 90, 95]
})

df2 = pd.DataFrame({
    "marks": [85, 99, 95]
})

result = df1.compare(df2)

print(result)
```
